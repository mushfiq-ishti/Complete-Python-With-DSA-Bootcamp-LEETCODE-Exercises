You are given an undirected graph with n nodes, numbered from 0 to n - 1. The graph is represented by a 2D integer array edges, where edges[i] = [ai, bi] denotes an undirected edge between the nodes ai and bi.

Your task is to return the number of pairs of different nodes that are unreachable from each other. A pair (i, j) is considered unreachable if there is no path from node i to node j.
Input:
n: An integer representing the number of nodes.
edges: A 2D list of integer pairs representing the edges between nodes.
Output:
Return the number of unreachable pairs of nodes.
Example:
Input:
n = 5
edges = [[0, 1], [0, 2], [3, 4]]
Output:
6
Explanation:
- The graph has two connected components: {0, 1, 2} and {3, 4}.
- Unreachable pairs are: (0, 3), (0, 4), (1, 3), (1, 4), (2, 3), (2, 4).
- Hence, the output is 6.
 Input:
n = 7
edges =[[0, 2], [0, 5], [2, 4], [1, 6], [5, 4]]
 Output:
14
Explanation:
- The graph has two connected components: {0, 2, 4, 5} and {1, 6}.
- Unreachable pairs are: (0, 1), (0, 6), (2, 1), (2, 6), (4, 1), (4, 6), (5, 1), (5, 6).
- Hence, the output is 14.

```python
def count_unreachable_pairs(n, edges):
    # Core logic for the learner to implement
    def edge_list_to_adj_matrix(n,edges):
        adj_matrix=[[0]*n for i in range(n)]
        for row in edges:
            source=row[0]
            dest=row[1]
            adj_matrix[source][dest]=1
            adj_matrix[dest][source]=1
        return adj_matrix
    
    
    def dfs_all_components(n,adj_matrix):

        def dfs_helper(adj_matrix,vertex,visited,result):
            result.append(vertex)
            visited[vertex]=True
            for neighbour in range(n):
                if neighbour==vertex:
                    continue
                if adj_matrix[vertex][neighbour]!=0:
                    if visited[neighbour]==False:
                        dfs_helper(adj_matrix,neighbour,visited,result)
            return result
    
        all_traversal=[]
        visited=[False]*n
        for start_vertex in range(n):
            if visited[start_vertex]==False:
                result=[]
                all_traversal.append(dfs_helper(adj_matrix,start_vertex,visited,result))
    
        return all_traversal

    adj_matrix=edge_list_to_adj_matrix(n,edges)
    all_componet_traversal=dfs_all_components(n,adj_matrix)

    #return all_componet_traversal
    sum_of_unreachable_node=0
    for i in range(len(all_componet_traversal)-1):
        for j in range(i+1,len(all_componet_traversal)):
            first_traversal=all_componet_traversal[i]
            second_traversal=all_componet_traversal[j]
            unreachable_node=len(first_traversal)*len(second_traversal)
            sum_of_unreachable_node+=unreachable_node
            
    
    return sum_of_unreachable_node

count_unreachable_pairs(7,[[0,2],[0,5],[2,4],[1,6],[5,4]])
```

<h2><b>Question-2</b></h2>
You are given a directed graph of n nodes labeled from 0 to n - 1. The graph is represented by a 0-indexed 2D integer array graph where graph[i] is an integer array of nodes adjacent to node i, meaning there is an edge from node i to each node in graph[i].

A node is a terminal node if there are no outgoing edges. A node is a safe node if every possible path starting from that node leads to a terminal node (or another safe node).

Your task is to return an array containing all the safe nodes of the graph. The answer should be sorted in ascending order.
Input Parameters:
graph (List[List[int]]): Adjacency list representing the directed graph.

Output:
List[int]: A list of safe nodes sorted in ascending order.
Example:

Input: graph = [[1,2],[2,3],[5],[0],[5],[],[]]
Output: [2,4,5,6]
 
Input: graph = [[1,2,3,4],[1,2],[3,4],[0,4],[]]
Output: [4]

```python
def eventualSafeNodes(graph):
    """
    Function to find all the safe nodes in a directed graph.
    
    :param graph: List[List[int]] -> Adjacency list representing the directed graph
    :return: List[int] -> List of all safe nodes sorted in ascending order
    """
    # TODO: Implement the logic to find all safe nodes
    def dfs_cycle_finder(starting_vertex, graph, visited_vertices=None):
        if visited_vertices is None:
            visited_vertices = []
    
        if starting_vertex in visited_vertices:
            return True  # If the node is already visited, cycle is found
    
        visited_vertices.append(starting_vertex)
        neighbours = graph[starting_vertex]
    
        for neighbour in neighbours:  
            if dfs_cycle_finder(neighbour, graph, visited_vertices)==True:
                return True  # If any recursive call returns True, propagate it upwards
    
        #removing the vertices which were not in forming a cycle if cycle found later
        visited_vertices.remove(starting_vertex)  # Backtrack after exploring all paths. Useful only for undirected graphs. can be removed here also
        return False  # If no cycle found
        
    vertices=list(range(len(graph)))
    ans=[]
    for starting_vertex in vertices:
        if dfs_cycle_finder(starting_vertex,graph)==False:
            ans.append(starting_vertex)
    return ans

graph = [[1,2,3,4],[1,2],[3,4],[0,4],[]]
eventualSafeNodes(graph)
```

The tree is represented using an array of edges where each edge connects two nodes in the tree.

Input:

n (integer): The number of nodes in the tree.

edges (List[List[int]]): A 2D list of edges, where edges[i] = [u, v] represents an edge between nodes u and v.

Output:

Return a list of integers where the i-th element is the number of nodes in the subtree rooted at node i.

Examples:

Input: 
n = 6
edges = [[0,1],[0,2],[1,3],[1,4],[2,5]]
Output: [6, 3, 2, 1, 1, 1]
Explanation: 
Node 0 is the root and has 5 other nodes in its subtree.

Node 2 has 2 nodes (including itself).
Nodes 3, 4, and 5 are leaves and have 1 node each (themselves).
 
Input: 
n = 5
edges = [[0,1],[0,2],[2,3],[2,4]]
Output: [5, 1, 3, 1, 1]
Explanation:
Node 0 has 5 nodes in total (entire tree).
Node 1 is a leaf and has only itself.
Node 2 has 3 nodes in its subtree.
Nodes 3 and 4 are leaves with 1 node each.

```python
def countSubtreeNodes(n, edges):
    # To be implemented by the student
    def edge_list_to_adjecency_list(n,edges):
        adjecency_list=[[] for i in range(n)]
        for edge in edges:
            source=edge[0]
            dest=edge[1]
            adjecency_list[source].append(dest)
        return adjecency_list

    def dfs_finding_nodes_in_subtrees(adjecency_list,starting_vertex,result=None):
        if result is None:
            result=[]
        result.append(starting_vertex)
        neighbours=adjecency_list[starting_vertex]
        for neighbour in neighbours:
            dfs_finding_nodes_in_subtrees(adjecency_list,neighbour,result)
        return result

    adj_list=edge_list_to_adjecency_list(n,edges)
    ans=[]
    for i in range(n):
        ans.append(len(dfs_finding_nodes_in_subtrees(adj_list,i)))
    return ans

#Example of how the input could look
n = 6
edges = [[0,1],[0,2],[1,3],[1,4],[2,5]]
countSubtreeNodes(n,edges)
```

Explanation:
- There are two connected components: {0, 1} and {2, 3}, but neither is complete.
- Hence, the output is 0.

Hint: A connected component is complete if and only if the number of edges in the component is equal to m*(m-1)/2, where m is the number of nodes in the component.

```python
def count_complete_components(n, edges):
    # Core logic for the learner to implement
    def common_elem_in_2_list(l1,l2):
        #probably could be dine using something built in but dont know at this moment
        flag=False
        for item in l1:
            if item in l2:
                flag=True
                break
        return flag
    def edge_list_to_adj_matrix(n,edges):
        adj_matrix=[[0]*n for i in range(n)]
        for row in edges:
            source=row[0]
            dest=row[1]
            adj_matrix[source][dest]=1
            adj_matrix[dest][source]=1
        return adj_matrix
    
    
    def dfs_all_components(n,adj_matrix):

        def dfs_helper(adj_matrix,vertex,visited,result):
            result.append(vertex)
            visited[vertex]=True
            for neighbour in range(n):
                if neighbour==vertex:
                    continue
                if adj_matrix[vertex][neighbour]!=0:
                    if visited[neighbour]==False:
                        dfs_helper(adj_matrix,neighbour,visited,result)
            return result
    
        all_traversal=[]
        visited=[False]*n
        for start_vertex in range(n):
            if visited[start_vertex]==False:
                result=[]
                all_traversal.append(dfs_helper(adj_matrix,start_vertex,visited,result))
    
        return all_traversal
        
    adj_matrix=edge_list_to_adj_matrix(n,edges)
    all_componet_traversal=dfs_all_components(n,adj_matrix)
    #return all_componet_traversal

    #finding no of edges in each traversal:
    no_of_complete_graph=0
    for traversal in all_componet_traversal:
        if len(traversal)==1:
            no_of_complete_graph+=1
            continue
        no_of_edges_in_traversal=0 #how many edges in the traversal
        for edge in edges:
            if common_elem_in_2_list(traversal,edge):
                no_of_edges_in_traversal+=1
        m=len(traversal)
        if no_of_edges_in_traversal==(m*(m-1)/2):
            no_of_complete_graph+=1
        
    return no_of_complete_graph
    
n = 6
edges = [[0,1],[0,2],[1,2],[3,4]]
count_complete_components(n,edges)
```

The graph is represented as an adjacency list where graph[i] contains a list of all nodes you can visit from node i (i.e., there is a directed edge from node i to each node in graph[i]). The graph is guaranteed to be a DAG. Return all possible paths from node 0 to node n - 1 in any order.

Input Parameters:
graph (List[List[int]]): An adjacency list where graph[i] represents the nodes that can be reached from node i.
Output:
Return a list of all possible paths from node 0 to node n - 1.

Example:

Input: graph = [[1,2],[3],[3],[]]
Output: [[0,1,3],[0,2,3]]
Explanation: There are two possible paths from node 0 to node 3.
 
Input: graph = [[4,3,1],[3,2,4],[3],[4],[]]
Output: [[0,4],[0,3,4],[0,1,4],[0,1,3,4],[0,2,3,4]]
Explanation: All possible paths from node 0 to node 4

```python
def all_paths_source_target(graph):
    """
    Function to find all possible paths from node 0 to node n - 1 in a DAG.
    
    :param graph: List[List[int]] -> Adjacency list representing the DAG
    :return: List[List[int]] -> List of all possible paths from node 0 to node n - 1
    """
    # TODO: Implement the logic to find all paths from source to target
    def dfs_all_path_source_to_target(starting_vertex,graph,path=None,result=None):
        if path is None:
            path=[]
        if result is None:
            result=[]
            
        path.append(starting_vertex)
        if starting_vertex==len(graph)-1:
            result.append(path.copy())
            return
        neighbours=graph[starting_vertex]
        for neighbour in neighbours:
            dfs_all_path_source_to_target(neighbour,graph,path,result)
            if len(path)!=0:
                path.pop()
        return result

    return dfs_all_path_source_to_target(0,graph)
    
graph = [[4,3,1],[3,2,4],[3],[4],[]]
all_paths_source_target(graph)
```

Your goal is to determine whether you can visit all the rooms, starting from room 0.

Input:

rooms: A list of lists where rooms[i] is a list of keys you can collect in room i.

Output:

Return True if you can visit all the rooms, otherwise return False.

Example:

Input:
rooms = [[1], [2], [3], []]
 
Output:
True
 
Explanation:
- You start in room 0 and collect key 1.
- You move to room 1 and collect key 2.
- You move to room 2 and collect key 3.
- You move to room 3, and all rooms are visited.
 
Input:
rooms = [[1, 3], [3, 0, 1], [2], [0]]
 
Output:
False
 
Explanation:
- You start in room 0 and collect keys 1 and 3.
- Room 2 is never visited because you do not have its key.

```python
def can_visit_all_rooms(rooms):
    # Core logic for the learner to implement
    #rooms here basically an adjecncy list for directed graph. Let us run a dfs from 0 with tracking visited list
    visited=[False]*len(rooms)
    def dfs_all_rooms(starting_vertex,adjency_list,visited,result=None):
        if result is None:
            result=[]
        if visited[starting_vertex]==False:
            result.append(starting_vertex)
            visited[starting_vertex]=True
        else:
            return
        neighbour_rooms=adjency_list[starting_vertex]
        for room in neighbour_rooms:
            dfs_all_rooms(room,adjency_list,visited,result)
        return result
        
    visited_rooms_from_0=dfs_all_rooms(0,rooms,visited)
    return len(rooms)==len(visited_rooms_from_0)

rooms = [[1,3],[3,0,1],[2],[0]]
can_visit_all_rooms(rooms)
```
