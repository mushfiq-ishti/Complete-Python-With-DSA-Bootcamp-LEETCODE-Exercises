```python
heart_y.head()
```

```python
sc.fit(heart_xtrain)
```

```python
#preserving the column names
heart_trainColumnNames=heart_xtrain.columns.values
heart_trainColumnNames
```

```python
heart_xtrain_sc= sc.fit_transform(heart_xtrain)
heart_xtrain_sc
```

```python
heart_xtest_sc= sc.transform(heart_xtest)
heart_xtest_sc
```

```python
#restoring column names
heart_xtrain_sc=pd.DataFrame(heart_xtrain_sc,columns=heart_trainColumnNames)
heart_xtrain_sc.head()
```

```python
heart_xtest_sc=pd.DataFrame(heart_xtest_sc,columns=heart_trainColumnNames)
heart_xtest_sc.head()
```

**Training and Testing**

Using Logistic Regression

```python
lr = LogisticRegression()
lr.fit(heart_xtrain_sc, heart_ytrain)
```

```python
print('training:', lr.score(heart_xtrain_sc, heart_ytrain))
print('testing:', lr.score(heart_xtest, heart_ytest))
```

Using Decesion tree

```python
dt= DecisionTreeClassifier(random_state=25,ccp_alpha=0.001,max_features='sqrt',max_depth=6,min_samples_leaf=3)
dt.fit(heart_xtrain_sc, heart_ytrain)
```

```python
print('training:', dt.score(heart_xtrain_sc, heart_ytrain))
print('testing:', dt.score(heart_xtest, heart_ytest))
```

# Regression Model on Housing dataset

**Performing EDA:**

```python
housing_data.info()
```

```python
housing_data.describe()
```

```python
housing_data.nunique()
```

**Preprocessig the data:**

```python
#Mapping the categorical feature
housing_data['mainroad'] = housing_data['mainroad'].map( {'yes': 1, 'no': 0} )
housing_data['guestroom'] = housing_data['guestroom'].map( {'yes': 1, 'no': 0} )
housing_data['basement'] = housing_data['basement'].map( {'yes': 1, 'no': 0} )
housing_data['hotwaterheating'] = housing_data['hotwaterheating'].map( {'yes': 1, 'no': 0} )
housing_data['airconditioning'] = housing_data['airconditioning'].map( {'yes': 1, 'no': 0} )
housing_data['prefarea'] = housing_data['prefarea'].map( {'yes': 1, 'no': 0} )
housing_data['furnishingstatus'] = housing_data['furnishingstatus'].map( {'furnished': 1, 'semi-furnished': 2, 'unfurnished': 3} )

housing_data.head()
```

```python
#Feature Engineering
housing_data['total_rooms']=housing_data['bedrooms']+housing_data['bathrooms']+housing_data['guestroom']
housing_data.head()
```

```python
housing_data=housing_data.drop('bedrooms',axis=1)
housing_data=housing_data.drop('bathrooms',axis=1)
housing_data=housing_data.drop('guestroom',axis=1)
housing_data.head()
```

**Train-Test Split:**

```python
#splitting target and feature variable
house_x=housing_data.drop('price',axis=1)
house_y=housing_data['price']
```

```python
#checking feature and target dataframes
house_x.head()
```

```python
house_y.head()
```

```python
house_xtrain, house_xtest, house_ytrain, house_ytest = train_test_split(house_x,house_y, test_size=0.2, random_state=25)
```

**Feature Scaling:**

```python
sc= StandardScaler()
```

```python
sc.fit(house_xtrain)
```

```python
#preserving the column names
housing_trainColumnNames=house_xtrain.columns.values
housing_trainColumnNames
```

```python
house_xtrain_sc= sc.fit_transform(house_xtrain)
house_xtrain_sc
```

```python
house_xtest_sc= sc.transform(house_xtest)
house_xtest_sc
```

```python
#restoring column names
house_xtrain_sc=pd.DataFrame(house_xtrain_sc,columns=housing_trainColumnNames)
house_xtrain_sc.head()
```

```python
house_xtest_sc=pd.DataFrame(house_xtest_sc,columns=housing_trainColumnNames)
house_xtest_sc.head()
```

**Training and Testing**

Using simple Linear Regression:

```python
from sklearn.linear_model import LinearRegression
# Creating a Linear Regression model
model = LinearRegression()

# Training the model using the training data
model.fit(house_xtrain_sc, house_ytrain)
```

```python
# Prediction on the test data
house_ypred = model.predict(house_xtest_sc)
```

```python
#model Evaluation
# Calculate Mean Squared Error (MSE)
mse = mean_squared_error(house_ytest, house_ypred)

# Calculate R-squared (R2)
r2 = r2_score(house_ytest, house_ypred)

print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
```
