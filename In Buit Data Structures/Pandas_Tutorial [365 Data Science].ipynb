# Methods vs Functions:
Function is independent entity or not associated with any object
Method is asspciated with a specific class when called or invoked. It can manipulate object's state. Each libraty has specific methods and only they can be used for the object

```python
start_date_deposits= pd.Series({
    '7/4/2014':24450,
    '1/2/2015':28900,
    '12/8/2012':43240,
    '2/20/2015':32450
})
print(start_date_deposits)
```

```python
#These methods are used both for Pandas and NumPy
print(start_date_deposits.sum()) #summing all the elements
print(start_date_deposits.min()) #minimum value
print(start_date_deposits.max()) #maximum value
print(start_date_deposits.idxmax()) #index of maximum value
print(start_date_deposits.idxmin()) #index of minimum value
```

```python
#These methods are used for Pandas only
print(start_date_deposits.head()) #gives first five elements
print((start_date_deposits.tail())) #gives last five elements
```

## Panda DataFrame
Multi-Column Data representing different variable and datatype. We need two index here: one for row index and another column index. Single index will retuen only entire row or entire column.
We can provide a whole object that contains the values of an entire column to the dictionary keys.
Let's crete a dataframe from dictionary:

#Creating dataframe from single dictionary

```python
data={'Productname':['Product A','Product B','Product C'],'ProductPrice':[22250,16600,12500]}
print(data)
df=pd.DataFrame(data) #default implicit indexing
print(df)
```

```python
#adding explicit labels: (label lengths must be same as no of elements)
df=pd.DataFrame(data,index=['A','B','C'])
print(df)
```

```python
#adding index from another list
product_IDs=['A','B','C']
df= pd.DataFrame(data,index=product_IDs)
print(df)
```

#Creating dataframe of list of dictionaries:

```python
#all columns will need data
data=[{'ProductName':'Product A','ProductPrice':22250},{'ProductName':'Product B','ProductPrice':16600},{'ProductName':'Product C','ProductPrice':12500}]
df=pd.DataFrame(data)
print(df)
```

#Creating Dataframe from Pandas Series:

```python
ser_products=pd.Series(['Product A','Product B','Product C'])
ser_prices=pd.Series([22250,16600,12500])

data={'ProductName':ser_products,'ProductPrice':ser_prices}
df=pd.DataFrame(data)
print(df)
```

```python
# Adding explicit index to this type of creation is bit different:
ser_products=pd.Series(['Product A','Product B','Product C'],index=['A','B','C'])
ser_prices=pd.Series([22250,16600,12500],index=['A','B','C'])

data={'ProductName':ser_products,'ProductPrice':ser_prices}
df=pd.DataFrame(data)
print(df)

#changing the index sequence for any of the pandas series, will eventually rearrange with corresponding Product name
```

#Creating Dataframe from List of Lists

```python
data=[
    ['Product A',22250],
    ['Product B',16600],
    ['Product C',12500]
]
df=pd.DataFrame(data)
print(df)
print('\n')
#adding columns and indices
df.columns=['ProductName','ProductPrice']
df.index=['A','B','C']
print(df)
```

#The Professional way of creating DataFrame

```python
df=pd.DataFrame(data=[
    ['Product A',22250],
    ['Product B',16600],
    ['Product C',12500]
],
columns=['ProductName','ProductPrice'],
index=['A','B','C'])

print(df)
print(df.shape)
```

#Data selction in Pandas dataFrame
Extracts certain data from full dataset

```python
data=pd.read_csv('Lending-company.csv',index_col='LoanID')
lending_co_data=data.copy()
```

```python
lending_co_data.Product #will return the product column
lending_co_data['Product'] #will return the product column using indexing operator. More used
lending_co_data=[['Product','Location']] #will return the product and location column. Needs two bracktes

#more good way:
prod_loc=['Product','Location']
lending_co_data[prod_loc].head()
```

#iloc[ ] indexers:
Used as accessor.

```python
lending_co_data.iloc[1] #will return 2nd row
lending_co_data.iloc[1,3] #will return 4th info of 2nd row
```

```python
lending_co_data.iloc[1:4,:] #will return 2nd to 4th row
```

#.loc[ ] accessor:
sub select information from a DataFrame by referring to its index labels

```python
lending_co_data.loc['LoadID_3',:] # will return entire row of load ID 3
lending_co_data.loc['Load_ID3','Location']    #will return interest rate of load ID 3
lending_co_data.loc[:,'Location'] #will return all the locations
```
